# GPU for faster computation

```python

import torch
import torch_directml

# 1. Get the AMD GPU as a DirectML device
device = torch_directml.device()
print("Using device:", device)   # e.g. privateuseone:0

# 2. Move tensors to the device
#    For x @ y the inner dims must match: (100 x 255) @ (255 x 50)
x = torch.randn(100, 255).to(device)
y = torch.randn(255, 50).to(device)


```

```python
%%time
# 3. Operation runs on the AMD GPU
z = x @ y
print("Result shape:", z.shape)   # (100, 50)
print("Tensor is on:", z.device)
```
